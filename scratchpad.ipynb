```
import os
import subprocess

print("=" * 80)
print("1. RUNNING CPU (WITHOUT GPU) TEST SUITE")
print("=" * 80)
os.system("source $HOME/.cargo/env 2>/dev/null || true; cd /content/ultra_performance_kernel/crates/shelf_db && cargo test --test test_cpu_without_gpu -- --nocapture")

print("\n" + "=" * 80)
print("2. RUNNING PHYSICAL GPU SILICON TEST SUITE")
print("=" * 80)
os.system("source $HOME/.cargo/env 2>/dev/null || true; export RUSTFLAGS='-L /usr/local/lib'; export LD_LIBRARY_PATH=/usr/local/lib:$LD_LIBRARY_PATH; cd /content/ultra_performance_kernel/crates/shelf_db && cargo test --features gpu --test test_gpu_silicon -- --nocapture")

print("\n" + "=" * 80)
print("3. MULTI-STACK HARDWARE BENCHMARK: ULTRA CONVEYOR KERNEL VS NATIVE CUDA C++")
print("=" * 80)

bench_code = '''
import ctypes
import numpy as np
import time

lib = ctypes.CDLL('/usr/local/lib/libsheaf_gpu.so')
lib.create_sheaf_session.argtypes = [ctypes.c_int, ctypes.c_int, ctypes.c_int, ctypes.c_int, ctypes.c_int, ctypes.c_float, ctypes.POINTER(ctypes.c_int), ctypes.POINTER(ctypes.c_int), ctypes.POINTER(ctypes.c_float), ctypes.POINTER(ctypes.c_float)]
lib.create_sheaf_session.restype = ctypes.c_void_p
lib.execute_sheaf_cuda_graph.argtypes = [ctypes.c_void_p]
lib.execute_sheaf_cuda_graph.restype = ctypes.c_int
lib.execute_sheaf_iterative.argtypes = [ctypes.c_void_p]
lib.execute_sheaf_iterative.restype = ctypes.c_int
lib.get_sheaf_output.argtypes = [ctypes.c_void_p, ctypes.POINTER(ctypes.c_float)]
lib.get_sheaf_output.restype = ctypes.c_int
lib.get_sheaf_dirichlet_energy.argtypes = [ctypes.c_void_p]
lib.get_sheaf_dirichlet_energy.restype = ctypes.c_float
lib.reset_sheaf_input.argtypes = [ctypes.c_void_p, ctypes.POINTER(ctypes.c_float)]
lib.reset_sheaf_input.restype = ctypes.c_int
lib.destroy_sheaf_session.argtypes = [ctypes.c_void_p]
lib.destroy_sheaf_session.restype = None

NUM_NODES = 256
DEGREE = 6
STALK_DIM = 4
CHANNELS = 32
NUM_LAYERS = 100
TAU = 0.15

row_ptr = [0]
col_ind = []
for i in range(NUM_NODES):
    for d in range(1, DEGREE + 1):
        col_ind.append((i + d) % NUM_NODES)
    row_ptr.append(len(col_ind))
num_edges = len(col_ind)

rho = []
for e in range(num_edges):
    for i in range(STALK_DIM):
        for j in range(STALK_DIM):
            rho.append(1.0 if i == j else (0.05 if (i + j + e) % 7 == 0 else 0.0))

total_elements = NUM_NODES * STALK_DIM * CHANNELS
x_in = [float(np.sin(i * 0.017)) for i in range(total_elements)]

c_row_ptr = (ctypes.c_int * len(row_ptr))(*row_ptr)
c_col_ind = (ctypes.c_int * len(col_ind))(*col_ind)
c_rho = (ctypes.c_float * len(rho))(*rho)
c_x_in = (ctypes.c_float * len(x_in))(*x_in)

handle = lib.create_sheaf_session(NUM_NODES, num_edges, STALK_DIM, CHANNELS, NUM_LAYERS, TAU, c_row_ptr, c_col_ind, c_rho, c_x_in)

# 1. First iteration cold-start
lib.reset_sheaf_input(handle, c_x_in)
t_cold_start = time.perf_counter()
lib.execute_sheaf_cuda_graph(handle)
cold_start_ms = (time.perf_counter() - t_cold_start) * 1000.0

# 2. Native CUDA iterative benchmark (50 runs of 100 layers)
latencies_iter = []
for _ in range(50):
    lib.reset_sheaf_input(handle, c_x_in)
    t0 = time.perf_counter()
    lib.execute_sheaf_iterative(handle)
    latencies_iter.append((time.perf_counter() - t0) * 1000.0)

# 3. Conveyor Belt CUDA Graph benchmark (50 runs of 100 layers)
latencies_graph = []
for _ in range(50):
    lib.reset_sheaf_input(handle, c_x_in)
    t0 = time.perf_counter()
    lib.execute_sheaf_cuda_graph(handle)
    latencies_graph.append((time.perf_counter() - t0) * 1000.0)

energy = lib.get_sheaf_dirichlet_energy(handle)

# Memory footprint calculation
vram_allocated_bytes = (
    total_elements * 4 * 2 + # d_x_a, d_x_b
    num_edges * STALK_DIM * STALK_DIM * 4 + # d_rho
    (NUM_NODES + 1) * 4 + # d_row_ptr
    num_edges * 4 + # d_col_ind
    1 * 4 # d_energy
)

lib.destroy_sheaf_session(handle)

p_iter = np.median(latencies_iter) / NUM_LAYERS * 1000.0
p_graph = np.median(latencies_graph) / NUM_LAYERS * 1000.0
speedup = p_iter / p_graph

print('=' * 80)
print('ULTRA PERFORMANCE KERNEL: PHYSICAL TESLA T4 SILICON TELEMETRY')
print('=' * 80)
print(f'VRAM Allocated Footprint:           {vram_allocated_bytes:,} bytes (Pinned GDDR6)')
print(f'First Iteration Cold-Start:         {cold_start_ms:.3f} ms')
print(f'Native CUDA Iterative Step Latency: {p_iter:.2f} us / step')
print(f'Conveyor Belt CUDA Graph Latency:   {p_graph:.2f} us / step')
print(f'Throughput (Graph Conveyor Engine): {1_000_000.0 / p_graph:,.1f} diffusion steps / sec')
print(f'Speedup vs Native CUDA:             {speedup:.2f}x FASTER')
print(f'Terminal Dirichlet Energy:          {energy:.6f}')
print('=' * 80)
'''

with open('/content/run_bench.py', 'w') as f:
    f.write(bench_code)

os.system("python3 /content/run_bench.py")

print("\n" + "=" * 80)
print("4. PACKAGING AND DOWNLOADING EVIDENCE ARTIFACTS")
print("=" * 80)
import shutil
from google.colab import files
shutil.make_archive('/content/ultra_performance_kernel_evidence', 'zip', '/content/ultra_performance_kernel')
print("Artifact created at /content/ultra_performance_kernel_evidence.zip")
files.download('/content/ultra_performance_kernel_evidence.zip')
print("Verification and benchmarking run complete!")
```

```
import subprocess

def run_cmd(title, cmd):
    print("=" * 80)
    print(title)
    print("=" * 80)
    res = subprocess.run(cmd, shell=True, capture_output=True, text=True, executable='/bin/bash')
    if res.stdout:
        print(res.stdout)
    if res.stderr:
        print("STDERR:\n", res.stderr)
    return res.returncode

# 1. CPU Tests
run_cmd("1. CPU (WITHOUT GPU) INTEGRATION TESTS",
        "export PATH=$HOME/.cargo/bin:$PATH; cd /content/ultra_performance_kernel/crates/shelf_db && cargo test --test test_cpu_without_gpu -- --nocapture")

# 2. Physical GPU Silicon Tests
run_cmd("2. PHYSICAL GPU SILICON INTEGRATION TESTS (TESLA T4)",
        "export PATH=$HOME/.cargo/bin:$PATH; export RUSTFLAGS='-L /usr/local/lib'; export LD_LIBRARY_PATH=/usr/local/lib:$LD_LIBRARY_PATH; cd /content/ultra_performance_kernel/crates/shelf_db && cargo test --features gpu --test test_gpu_silicon -- --nocapture")

# 3. Python Multi-Stack Benchmark matching backend.rs C-ABI
bench_py = '''
import ctypes
import numpy as np
import time

lib = ctypes.CDLL('/usr/local/lib/libsheaf_gpu.so')

# C-ABI definitions matching backend.rs
lib.init_sheaf_gpu_vram.argtypes = [
    ctypes.c_int, ctypes.c_int, ctypes.c_int, ctypes.c_int,
    ctypes.POINTER(ctypes.c_int), ctypes.POINTER(ctypes.c_int),
    ctypes.POINTER(ctypes.c_float), ctypes.POINTER(ctypes.c_float),
    ctypes.POINTER(ctypes.c_void_p), ctypes.POINTER(ctypes.c_size_t)
]
lib.init_sheaf_gpu_vram.restype = ctypes.c_int

lib.reset_sheaf_input.argtypes = [ctypes.c_void_p, ctypes.POINTER(ctypes.c_float)]
lib.reset_sheaf_input.restype = ctypes.c_int

lib.run_sheaf_conveyor_gpu_steps.argtypes = [
    ctypes.c_void_p, ctypes.c_int, ctypes.c_float,
    ctypes.POINTER(ctypes.c_float), ctypes.POINTER(ctypes.c_float)
]
lib.run_sheaf_conveyor_gpu_steps.restype = ctypes.c_int

lib.instantiate_sheaf_cuda_graph.argtypes = [ctypes.c_void_p, ctypes.c_int, ctypes.c_float]
lib.instantiate_sheaf_cuda_graph.restype = ctypes.c_int

lib.run_sheaf_cuda_graph.argtypes = [
    ctypes.c_void_p, ctypes.POINTER(ctypes.c_float), ctypes.POINTER(ctypes.c_float)
]
lib.run_sheaf_cuda_graph.restype = ctypes.c_int

lib.free_sheaf_gpu_vram.argtypes = [ctypes.c_void_p]
lib.free_sheaf_gpu_vram.restype = ctypes.c_int

NUM_NODES = 256
DEGREE = 6
STALK_DIM = 4
CHANNELS = 32
NUM_LAYERS = 100
TAU = 0.15

row_ptr = [0]
col_ind = []
for i in range(NUM_NODES):
    for d in range(1, DEGREE + 1):
        col_ind.append((i + d) % NUM_NODES)
    row_ptr.append(len(col_ind))
num_edges = len(col_ind)

rho = []
for e in range(num_edges):
    for i in range(STALK_DIM):
        for j in range(STALK_DIM):
            rho.append(1.0 if i == j else (0.05 if (i + j + e) % 7 == 0 else 0.0))

total_elements = NUM_NODES * STALK_DIM * CHANNELS
x_in = [float(np.sin(i * 0.017)) for i in range(total_elements)]

c_row_ptr = (ctypes.c_int * len(row_ptr))(*row_ptr)
c_col_ind = (ctypes.c_int * len(col_ind))(*col_ind)
c_rho = (ctypes.c_float * len(rho))(*rho)
c_x_in = (ctypes.c_float * len(x_in))(*x_in)

handle = ctypes.c_void_p()
vram_bytes = ctypes.c_size_t()

err = lib.init_sheaf_gpu_vram(
    NUM_NODES, num_edges, STALK_DIM, CHANNELS,
    c_row_ptr, c_col_ind, c_rho, c_x_in,
    ctypes.byref(handle), ctypes.byref(vram_bytes)
)
assert err == 0, f"init_sheaf_gpu_vram failed with code {err}"

# 1. Warm-up and instantiate CUDA Graph
lib.instantiate_sheaf_cuda_graph(handle, NUM_LAYERS, TAU)

# 2. Native CUDA iterative baseline (50 runs)
time_ms = ctypes.c_float()
dirichlet = ctypes.c_float()
latencies_iter = []
for _ in range(50):
    lib.reset_sheaf_input(handle, c_x_in)
    lib.run_sheaf_conveyor_gpu_steps(handle, NUM_LAYERS, TAU, ctypes.byref(time_ms), ctypes.byref(dirichlet))
    latencies_iter.append(time_ms.value)

# 3. Conveyor Belt CUDA Graph engine (50 runs)
latencies_graph = []
for _ in range(50):
    lib.reset_sheaf_input(handle, c_x_in)
    lib.run_sheaf_cuda_graph(handle, ctypes.byref(time_ms), ctypes.byref(dirichlet))
    latencies_graph.append(time_ms.value)

lib.free_sheaf_gpu_vram(handle)

p_iter = np.median(latencies_iter) / NUM_LAYERS * 1000.0 # us / step
p_graph = np.median(latencies_graph) / NUM_LAYERS * 1000.0 # us / step
speedup = p_iter / p_graph

print("=" * 80)
print("ULTRA PERFORMANCE CONVEYOR KERNEL: PHYSICAL TESLA T4 SILICON RESULTS")
print("=" * 80)
print(f"Allocated GDDR6 VRAM:               {vram_bytes.value:,} bytes")
print(f"Native CUDA Iterative Step Latency: {p_iter:.2f} us / step")
print(f"Conveyor Belt CUDA Graph Latency:   {p_graph:.2f} us / step")
print(f"Throughput (Graph Conveyor Engine): {1_000_000.0 / p_graph:,.1f} diffusion steps / sec")
print(f"Hardware Speedup vs Native CUDA:    {speedup:.2f}x FASTER THAN NATIVE CUDA")
print(f"Terminal Dirichlet Energy:          {dirichlet.value:.6f}")
print("=" * 80)
'''

with open('/content/bench_final.py', 'w') as f:
    f.write(bench_py)

run_cmd("3. MULTI-STACK HARDWARE BENCHMARK RESULTS", "python3 /content/bench_final.py")
```
